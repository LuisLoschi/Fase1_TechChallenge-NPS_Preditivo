## 1.1 O que é o NPS, como ele é medido e por que o NPS é importante?

Com base nesse problema, para seguirmos, primeiro precisamos entender: **O que é o NPS, como ele é medido e por que ele é importante?**

O Net Promoter Score é uma métrica de fidelidade e satisfação do cliente que mede a probabilidade de recomendação da sua empresa, produto ou serviço.

Ele é categorizado em três tipos:

🟢 Promotores (9-10): Clientes satisfeitos que indicam a marca.

🟡 Neutros (7-8): Clientes passivos, propensos à concorrência.

🔴 Detratores (0-6): Clientes insatisfeitos, propensos a falar mal da empresa.

O cálculo é feito subtraindo a porcentagem de detratores da porcentagem de promotores (excluindo os neutros):

> NPS = % Promotores - % Detratores

Ele é um fator importante pois ajuda em tópicos como:

- Custo de Aquisição de Clientes (CAC): É mais barato manter um cliente do que conquistar um novo.

- Vantagem Competitiva: Em um mercado de produtos similares, a experiência de entrega, atendimento e pós-venda é um diferencial real.

- Validação de Processos: Pode revelar problemas que os números não mostram, como uma embalagem amassada ou uma experiência ruim que a logística padrão não percebe

## 1.2 Quais áreas poderiam se beneficiar desses insights

Um projeto de NPS preditivo deve gerar valor para múltiplas áreas, não apenas para Ciência de Dados.

| Área | Uso prático dos insights |
|---|---|
| Logística | Identificar atrasos, regiões e tentativas de entrega associadas à detratação |

A lógica do projeto é tratar os dados operacionais como um **sistema de alerta**: quanto mais cedo a empresa identifica o risco, maior a chance de intervir antes que a insatisfação seja consolidada na resposta de NPS.


## 1.3 Os Impactos do NPS

O NPS pode impactar em alguns pontos:

- **Taxa de Recompra e Ciclo de Vida (LTV):**
Clientes classificados como "Promotores" possuem um ciclo de vida maior e rentável, garantindo uma receita recorrente sem a necessidade de novos investimentos em mídia. No e-commerce, a margem de lucro real costuma se consolidar a partir da segunda ou terceira transação; portanto, um NPS baixo indica que a empresa está perdendo clientes prematuramente, tornando os investimentos em marketing ineficientes e incapazes de gerar crescimento sustentável.

- **Marketing Boca a Boca:**
O e-commerce é movido pela confiança. Clientes "Detratores" não apenas deixam de comprar, mas podem expor falhas em redes sociais e sites de reclamação, o que gera uma resistência maior em novos compradores e encarece o marketing. Por outro lado, os "Promotores" tornam-se um canal de marketing gratuito e de alta conversão: eles recomendam a marca espontaneamente, trazendo novos clientes com maior confiança.

- **Market Share:**
Embora a fatia de mercado possa ser expandida rapidamente através de promoções e escala logística, a liderança só é mantida através da fidelização. Empresas que sustentam um NPS elevado conseguem dominar o setor a longo prazo porque constroem comunidades de consumo sólidas. Essa lealdade cria uma barreira de proteção contra a concorrência, permitindo que a marca mantenha sua relevância mesmo diante de disputa de preços ou flutuações do mercado.

## 1.4 Indicadores que podem complementar a análise
Para melhorar o NPS, a empresa pode considerar aderir ou revisar alguns cenários:

- **CES (Customer Effort Score):** Mede o quão fácil foi para o cliente resolver um problema ou concluir a compra. Muitas vezes, um NPS baixo está ligado ao alto esforço.

- **OTD (On-Time Delivery):** Percentual de pedidos entregues exatamente no prazo. O atraso é o maior gerador de detratores no e-commerce.

- **First Contact Resolution (FCR):** Se o cliente precisou entrar em contato, o problema foi resolvido na primeira tentativa?

- **Churn Rate:** Cruzar a nota do NPS com o índice de abandono da base para validar a correlação entre insatisfação e perda real de receita.

- **SLA de Logística Inversa:** A facilidade de devolução é um fator crítico que pode transformar um potencial detrator em um promotor pela confiança gerada.

Assim, podemos usar os dados operacionais como um Sistema de Alerta: Se um pedido atrasou e o cliente já reclamou, a probabilidade de ele ser um detrator é alta. Agir antes de enviar a pesquisa é essencial para o crescimento da empresa.

# **3. Definição da variável target**

A variável-alvo principal é `nps_score`, uma nota de 0 a 10 coletada após a experiência de compra. Ela representa a satisfação percebida pelo cliente ao final da jornada e permite duas abordagens complementares:

1. **Regressão:** estimar a nota individual de NPS em escala contínua/discreta de 0 a 10.
2. **Classificação:** prever a categoria de relacionamento: Detrator, Neutro ou Promotor.

A classificação segue a regra clássica de NPS:

- **Detrator:** notas de 0 a 6;
- **Neutro:** notas 7 e 8;
- **Promotor:** notas 9 e 10.

Um ponto metodológico central é distinguir variáveis disponíveis **antes da pesquisa NPS** de variáveis coletadas simultaneamente à satisfação. A coluna `csat_internal_score` também mede satisfação e é coletada no mesmo contexto temporal do NPS. Portanto, pode ser analisada na EDA como referência comparativa, mas não deve ser usada como feature preditiva: em produção, ela não estará disponível no momento do scoring. Usá-la como entrada do modelo caracterizaria **data leakage**, inflando artificialmente as métricas e reduzindo a validade prática da solução.


## 3.1 Análise do NPS

Vamos analisar a variavel-alvo começando com a seguinte pergunta: **Qual a quantidade pessoas que são Promotoras, Neutras ou Detratoras?**

```python
# Classificação dos clientes em Detratores, Neutros e Promotores
def classify_nps_score(score: float) -> str:
    if score >= 9:
        return 'Promotor'
    if score >= 7:
        return "Neutro"
    return "Detrator"
```

```python
# Criar colunas auxiliares para análise de distribuição por categorias
working_df = dataset_nps.copy()
working_df["nps_category"] = working_df["nps_score"].apply(classify_nps_score)

working_df["age_group"] = pd.cut(
    working_df["customer_age"],
    bins=[0, 24, 34, 44, 54, 64, 120],
    labels=["Até 24", "25-34", "35-44", "45-54", "55-64", "65+"],
    include_lowest=True,
).astype(str)

working_df["tenure_group"] = pd.cut(
    working_df["customer_tenure_months"],
    bins=[-1, 6, 12, 24, 48, 1200],
    labels=["0-6m", "7-12m", "13-24m", "25-48m", "49m+"],
).astype(str)

category_order = ["Detrator", "Neutro", "Promotor"]
nps_distribution = working_df["nps_category"].value_counts().reindex(category_order, fill_value=0)
nps_percentual = (nps_distribution / len(working_df) * 100).round(2)
nps_score_global = nps_percentual["Promotor"] - nps_percentual["Detrator"]

pd.DataFrame({
    "quantidade": nps_distribution,
    "percentual_%": nps_percentual,
})
```

Aqui é possivel observar que aproximadamente 84% (2109) de todos os NPS são Detratores.

A menor presença de promotores reforça a necessidade de identificar os fatores que impactam negativamente a experiência do cliente

```python
print(f"NPS agregado da base: {nps_score_global:.2f}")
print(f"Nota média individual: {working_df['nps_score'].mean():.2f}")
print(f"Mediana da nota individual: {working_df['nps_score'].median():.2f}")
```

```python
nps_plot_df = (
    pd.DataFrame({
        "categoria": category_order,
        "clientes": nps_distribution.reindex(category_order).values,
        "percentual": nps_percentual.reindex(category_order).values,
    })
)

fig, ax = plt.subplots(figsize=(9.5, 5.4))
bars = sns.barplot(
    data=nps_plot_df,
    x="categoria",
    y="percentual",
    order=category_order,
    palette=CATEGORY_PALETTE,
    ax=ax,
)

for patch, row in zip(bars.patches, nps_plot_df.itertuples()):
    ax.annotate(
        f"{row.percentual:.1f}%\n{row.clientes:,} clientes".replace(",", "."),
        (patch.get_x() + patch.get_width() / 2, patch.get_height()),
        xytext=(0, 8),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=10,
        color="#1F2D3D",
    )

ax.axhline(50, color="#9AA6B2", linewidth=1, linestyle="--", alpha=0.55)
ax.text(2.42, 50.8, "referencia 50%", color="#66727F", fontsize=9, ha="right")
fig.suptitle("Base concentrada em detratores", x=0.01, y=0.99, ha="left", fontsize=16, weight="bold")
fig.text(
    0.01,
    0.92,
    f"NPS agregado = {nps_score_global:.1f} | nota media = {working_df['nps_score'].mean():.2f}",
    fontsize=11,
    color="#52616F",
)
ax.set_xlabel("")
ax.set_ylabel("Participacao na base (%)")
ax.set_ylim(0, max(nps_plot_df["percentual"]) * 1.25)
clean_axis(ax)
plt.tight_layout(rect=[0, 0, 1, 0.88])
show_save_close(fig, "01_distribuicao_nps.png")
```

### 4.1 Detecção e avaliação de outliers

Variáveis operacionais de experiência do cliente tendem a apresentar caudas longas à direita: a maioria dos clientes tem jornadas sem grandes fricções, mas um subconjunto concentra atrasos extremos, múltiplos contatos e reclamações acima do padrão. Antes de treinar um modelo preditivo, é importante identificar esses valores extremos, quantificá-los e decidir explicitamente se eles representam casos reais de operação ou erros de registro.

O critério adotado é o IQR (intervalo interquartil): valores abaixo de Q1 − 1,5×IQR ou acima de Q3 + 1,5×IQR são sinalizados como outliers. Para variáveis que só admitem valores não negativos (atrasos, contatos, reclamações), o limite inferior é truncado em zero.


```python
outlier_cols = [
    "delivery_delay_days",
    "customer_service_contacts",
    "complaints_count",
    "resolution_time_days",
    "freight_value",
    "order_value",
]

fig, axes = plt.subplots(2, 3, figsize=(16, 8.5))
axes = axes.flatten()
outlier_summary_rows = []

for ax, col in zip(axes, outlier_cols):
    q1 = working_df[col].quantile(0.25)
    q3 = working_df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = max(0, q1 - 1.5 * iqr)
    upper_bound = q3 + 1.5 * iqr
    n_outliers = ((working_df[col] < lower_bound) | (working_df[col] > upper_bound)).sum()
    pct_outliers = n_outliers / len(working_df) * 100

    sns.boxplot(
        data=working_df,
        x="nps_category",
        y=col,
        order=category_order,
        palette=CATEGORY_PALETTE,
        width=0.58,
        linewidth=1,
        fliersize=2.5,
        ax=ax,
    )
    ax.set_title(f"{LABEL_MAP.get(col, col)}", loc="left", fontsize=11, weight="bold", pad=20)
    ax.text(
        0,
        1.03,
        f"{n_outliers} outliers ({pct_outliers:.1f}% da base)",
        transform=ax.transAxes,
        fontsize=9,
        color="#66727F",
    )
    ax.set_xlabel("")
    ax.set_ylabel("")
    clean_axis(ax)

    outlier_summary_rows.append({
        "variavel": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "limite_inferior": lower_bound,
        "limite_superior": upper_bound,
        "n_outliers": n_outliers,
        "pct_outliers_%": round(pct_outliers, 2),
    })

fig.suptitle("Fricções extremas aparecem principalmente nas métricas operacionais", x=0.01, ha="left", fontsize=16, weight="bold")
fig.text(0.01, 0.94, "Boxplots por categoria NPS com contagem de outliers pelo criterio IQR.", fontsize=11, color="#52616F")
plt.tight_layout(rect=[0, 0, 1, 0.91])
show_save_close(fig, "03_outliers_por_categoria_nps.png")

outlier_summary_df = pd.DataFrame(outlier_summary_rows)
display(outlier_summary_df.round(3))
```

Os boxplots por categoria NPS permitem avaliar se os outliers estão concentrados em grupos específicos de satisfação. Na base analisada, observa-se baixa incidência de outliers em variáveis como atraso (0,68%) e reclamações (1,16%), enquanto contatos com atendimento apresentam maior dispersão (7,04%).

De forma geral, os outliers estão associados a experiências operacionais mais extremas e não a erros evidentes de registro, indicando que carregam sinal relevante sobre a insatisfação do cliente.

Diante disso, os outliers são mantidos nas análises e na modelagem, pois representam justamente os casos críticos que o modelo deve ser capaz de identificar.

### 4.2 Pontos de ruptura operacionais

Pontos de ruptura são limiares a partir dos quais a experiência muda de patamar. Para o negócio, eles são mais acionáveis do que médias gerais, pois ajudam a transformar análise em regra operacional: priorizar clientes com atraso, múltiplos contatos ou reclamações antes que a pesquisa NPS seja enviada.


```python
def summarize_nps_by_discrete_feature(df: pd.DataFrame, feature: str) -> pd.DataFrame:
    summary = (
        df.groupby(feature, observed=True)
        .agg(
            clientes=("nps_score", "size"),
            nps_medio=("nps_score", "mean"),
            perc_detratores=("nps_category", lambda values: (values == "Detrator").mean() * 100),
            perc_promotores=("nps_category", lambda values: (values == "Promotor").mean() * 100),
        )
        .reset_index()
        .sort_values(feature)
    )
    return summary.round(3)


rupture_features = ["delivery_delay_days", "customer_service_contacts", "complaints_count", "delivery_attempts"]
rupture_summaries = {}

fig, axes = plt.subplots(2, 2, figsize=(15.5, 8.5))
axes = axes.flatten()

for ax, feature in zip(axes, rupture_features):
    summary = summarize_nps_by_discrete_feature(working_df, feature)
    rupture_summaries[feature] = summary
    display(summary)

    feature_label = LABEL_MAP.get(feature, feature)
    sns.lineplot(
        data=summary,
        x=feature,
        y="nps_medio",
        marker="o",
        color=EXECUTIVE_COLORS["primary"],
        linewidth=2.2,
        ax=ax,
    )
    ax2 = ax.twinx()
    sns.lineplot(
        data=summary,
        x=feature,
        y="perc_detratores",
        marker="s",
        color=EXECUTIVE_COLORS["Detrator"],
        linewidth=1.8,
        alpha=0.85,
        ax=ax2,
    )
    ax.set_title(feature_label, loc="left", fontsize=11, weight="bold")
    ax.set_xlabel("")
    ax.set_ylabel("NPS medio", color=EXECUTIVE_COLORS["primary"])
    ax2.set_ylabel("% detratores", color=EXECUTIVE_COLORS["Detrator"])
    ax.tick_params(axis="y", labelcolor=EXECUTIVE_COLORS["primary"])
    ax2.tick_params(axis="y", labelcolor=EXECUTIVE_COLORS["Detrator"])
    clean_axis(ax, grid_axis="y")
    ax2.grid(False)

fig.suptitle("Pontos de ruptura: a experiencia cai quando a friccao aumenta", x=0.01, ha="left", fontsize=16, weight="bold")
fig.text(0.01, 0.94, "Linha azul: NPS medio | linha vermelha: percentual de detratores.", fontsize=11, color="#52616F")
plt.tight_layout(rect=[0, 0, 1, 0.91])
show_save_close(fig, "07_pontos_ruptura_operacionais.png")
```

```python
def first_rupture_point(
    df: pd.DataFrame,
    feature: str,
    baseline_value: int | float = 0,
    min_group_size: int = 30,
    nps_drop_threshold: float = 1.0,
) -> dict:
    summary = summarize_nps_by_discrete_feature(df, feature)
    summary = summary[summary["clientes"] >= min_group_size].copy()

    if baseline_value in set(summary[feature]):
        baseline_nps = float(summary.loc[summary[feature] == baseline_value, "nps_medio"].iloc[0])
    else:
        baseline_nps = float(summary["nps_medio"].iloc[0])

    summary["queda_vs_referencia"] = baseline_nps - summary["nps_medio"]
    candidates = summary[summary["queda_vs_referencia"] >= nps_drop_threshold]

    if candidates.empty:
        rupture_value = np.nan
        rupture_nps = np.nan
        rupture_drop = np.nan
    else:
        first = candidates.iloc[0]
        rupture_value = first[feature]
        rupture_nps = first["nps_medio"]
        rupture_drop = first["queda_vs_referencia"]

    return {
        "variavel": feature,
        "referencia": baseline_value,
        "nps_referencia": baseline_nps,
        "limiar_ruptura": rupture_value,
        "nps_no_limiar": rupture_nps,
        "queda_nps": rupture_drop,
        "criterio": f"queda >= {nps_drop_threshold} ponto(s), n >= {min_group_size}",
    }


rupture_table = pd.DataFrame([
    first_rupture_point(working_df, "delivery_delay_days", baseline_value=0),
    first_rupture_point(working_df, "customer_service_contacts", baseline_value=0),
    first_rupture_point(working_df, "complaints_count", baseline_value=0),
    first_rupture_point(working_df, "delivery_attempts", baseline_value=1),
])

rupture_table.round(3)

rupture_exec = rupture_table.copy()
rupture_exec["variavel_label"] = rupture_exec["variavel"].map(lambda value: LABEL_MAP.get(value, value))
rupture_exec["limiar_label"] = rupture_exec["limiar_ruptura"].map(lambda value: "sem ruptura" if pd.isna(value) else f">= {value:g}")

kpis = [
    ("NPS Agregado", f"{nps_score_global:.1f}%", "Promotores - detratores", EXECUTIVE_COLORS["Detrator"]),
    ("Detratores", f"{nps_percentual['Detrator']:.1f}%", f"{nps_distribution['Detrator']:,} clientes".replace(",", "."), EXECUTIVE_COLORS["Detrator"]),
    ("Promotores", f"{nps_percentual['Promotor']:.1f}%", f"{nps_distribution['Promotor']:,} clientes".replace(",", "."), EXECUTIVE_COLORS["Promotor"]),
    ("Ruptura critica", "1 dia", "atraso ja derruba NPS", EXECUTIVE_COLORS["primary"]),
]

fig = plt.figure(figsize=(16, 9))
grid = fig.add_gridspec(3, 4, height_ratios=[0.72, 1.45, 1.45], hspace=0.48, wspace=0.35)
fig.suptitle("Resumo executivo: NPS pressionado por friccoes operacionais", x=0.02, y=0.98, ha="left", fontsize=18, weight="bold")
fig.text(0.02, 0.925, "Leitura para banca: a base e majoritariamente detratora; atraso, reclamacoes e contatos explicam a queda de experiencia.", fontsize=11, color="#52616F")

for idx, (title, value, subtitle, color) in enumerate(kpis):
    ax_card = fig.add_subplot(grid[0, idx])
    ax_card.set_facecolor("#F7F9FC")
    for spine in ax_card.spines.values():
        spine.set_visible(False)
    ax_card.set_xticks([])
    ax_card.set_yticks([])
    ax_card.text(0.04, 0.78, title.upper(), transform=ax_card.transAxes, fontsize=8.5, color="#66727F", weight="bold")
    ax_card.text(0.04, 0.35, value, transform=ax_card.transAxes, fontsize=24, color=color, weight="bold")
    ax_card.text(0.04, 0.12, subtitle, transform=ax_card.transAxes, fontsize=9.5, color="#52616F")

ax_dist = fig.add_subplot(grid[1:, :2])
sns.barplot(
    data=nps_plot_df,
    x="categoria",
    y="percentual",
    order=category_order,
    palette=CATEGORY_PALETTE,
    ax=ax_dist,
)
for patch, row in zip(ax_dist.patches, nps_plot_df.itertuples()):
    ax_dist.annotate(
        f"{row.percentual:.1f}%",
        (patch.get_x() + patch.get_width() / 2, patch.get_height()),
        xytext=(0, 8),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=11,
        color="#1F2D3D",
        weight="bold",
    )
ax_dist.set_title("Composicao da base NPS", loc="left", fontsize=13, weight="bold", pad=20)
ax_dist.set_xlabel("")
ax_dist.set_ylabel("Participacao (%)")
ax_dist.set_ylim(0, 85)
clean_axis(ax_dist)

ax_drop = fig.add_subplot(grid[1, 2:])
plot_drop = rupture_exec.sort_values("queda_nps", ascending=False).dropna(subset=["queda_nps"])
sns.barplot(
    data=plot_drop,
    y="variavel_label",
    x="queda_nps",
    color=EXECUTIVE_COLORS["Detrator"],
    ax=ax_drop,
)
for patch, row in zip(ax_drop.patches, plot_drop.itertuples()):
    ax_drop.annotate(
        f"-{row.queda_nps:.1f} pts",
        (patch.get_width(), patch.get_y() + patch.get_height() / 2),
        xytext=(6, 0),
        textcoords="offset points",
        ha="left",
        va="center",
        fontsize=9.5,
        color="#1F2D3D",
    )
ax_drop.set_title("Queda de NPS no primeiro ponto de ruptura", loc="left", fontsize=13, weight="bold")
ax_drop.set_xlabel("Queda vs. referencia")
ax_drop.set_ylabel("")
clean_axis(ax_drop, grid_axis="x")

ax_table = fig.add_subplot(grid[2, 2:])
ax_table.axis("off")
driver_short = {
    "delivery_delay_days": "Atraso",
    "customer_service_contacts": "Atendimento",
    "complaints_count": "Reclamacoes",
    "delivery_attempts": "Tentativas",
}
table_data = rupture_exec[["variavel", "limiar_label", "nps_referencia", "nps_no_limiar"]].copy()
table_data["driver"] = table_data["variavel"].map(driver_short).fillna(table_data["variavel"])
table_data = table_data[["driver", "limiar_label", "nps_referencia", "nps_no_limiar"]]
table_data["nps_referencia"] = table_data["nps_referencia"].map(lambda value: f"{value:.1f}")
table_data["nps_no_limiar"] = table_data["nps_no_limiar"].map(lambda value: "-" if pd.isna(value) else f"{value:.1f}")
matplotlib_table = ax_table.table(
    cellText=table_data.values,
    colLabels=["Driver", "Ruptura", "NPS ref.", "NPS ruptura"],
    colWidths=[0.28, 0.24, 0.20, 0.28],
    loc="center",
    cellLoc="left",
    colLoc="left",
)
matplotlib_table.auto_set_font_size(False)
matplotlib_table.set_fontsize(9.0)
matplotlib_table.scale(1, 1.45)
for (row, col), cell in matplotlib_table.get_celld().items():
    cell.set_edgecolor("#D9DEE7")
    if row == 0:
        cell.set_facecolor("#1F4E79")
        cell.set_text_props(color="white", weight="bold")
    else:
        cell.set_facecolor("#F7F9FC" if row % 2 else "white")
ax_table.set_title("Limiares acionaveis para gestao da experiencia", loc="left", fontsize=13, weight="bold", pad=10)

show_save_close(fig, "00_resumo_executivo_nps.png")



```

# **5. Inferência estatística**

A inferência estatística é utilizada para validar se os padrões observados na análise exploratória são estatisticamente significativos ou fruto de variação aleatória.

Como o `nps_score` é discreto, limitado entre 0 e 10 e apresenta distribuição não normal nos subgrupos, são adotados testes não paramétricos.

Nesta seção, cada análise é apresentada em linguagem de negócio, com definição de hipóteses, resultado do teste e interpretação orientada à tomada de decisão.

### 5.1 Normalidade do NPS por subgrupo

- **H₀:** o NPS apresenta distribuição aproximadamente normal dentro dos subgrupos.  
- **H₁:** o NPS não apresenta distribuição normal em pelo menos parte dos subgrupos.  

A verificação de normalidade orienta a escolha dos testes estatísticos. Resultados com p-valores baixos indicam rejeição da hipótese de normalidade, justificando o uso de testes não paramétricos.

Essa abordagem é consistente com a natureza do NPS, que é uma variável discreta, limitada (0 a 10) e frequentemente concentrada em extremos.

```python
def normality_by_group(
    df: pd.DataFrame,
    group_col: str,
    value_col: str = "nps_score",
    max_shapiro_n: int = 5000,
) -> pd.DataFrame:
    rows = []
    for group, values in df.groupby(group_col, observed=True)[value_col]:
        sample = values.dropna()
        if len(sample) < 3:
            rows.append({
                "grupo": group,
                "n": len(sample),
                "teste": None,
                "estatistica": np.nan,
                "p_valor": np.nan,
            })
            continue

        if len(sample) <= max_shapiro_n:
            stat, p_value = stats.shapiro(sample)
            test_name = "Shapiro-Wilk"
        else:
            z_sample = (sample - sample.mean()) / sample.std(ddof=1)
            stat, p_value = stats.kstest(z_sample, "norm")
            test_name = "Kolmogorov-Smirnov"

        rows.append({
            "grupo": group,
            "n": len(sample),
            "teste": test_name,
            "estatistica": stat,
            "p_valor": p_value,
        })

    return pd.DataFrame(rows).sort_values("p_valor")


normality_region = normality_by_group(working_df, "customer_region")
normality_age = normality_by_group(working_df, "age_group")

display(normality_region.round(5))
display(normality_age.round(5))
```

Os resultados do teste de Shapiro-Wilk indicam p-valores próximos de zero em todos os subgrupos analisados (por região e faixa etária). Dessa forma, rejeita-se a hipótese de normalidade para o NPS.

Esse resultado confirma que a distribuição do NPS não segue comportamento normal, o que é consistente com sua natureza discreta e limitada.

Diante disso, a utilização de testes não paramétricos nas análises subsequentes é justificada e necessária.

### 5.2 Testes de diferença de NPS por região e faixa etária

Nesta etapa, aplica-se o teste de Kruskal-Wallis para avaliar se há diferença estatisticamente significativa na distribuição do NPS entre regiões e faixas etárias.

Como o NPS não apresenta distribuição normal nos subgrupos, o Kruskal-Wallis é mais adequado do que uma ANOVA tradicional.

#### Região geográfica

- **H₀:** as regiões apresentam distribuição de NPS semelhante.
- **H₁:** ao menos uma região apresenta distribuição de NPS diferente.

#### Faixa etária

- **H₀:** as faixas etárias apresentam distribuição de NPS semelhante.
- **H₁:** ao menos uma faixa etária apresenta distribuição de NPS diferente.

Caso o teste global seja significativo, aplica-se o pós-hoc de Dunn para identificar quais grupos diferem entre si. Além do p-valor, calcula-se o eta-quadrado para avaliar a magnitude prática do efeito.

```python
def kruskal_test(df: pd.DataFrame, group_col: str, value_col: str = "nps_score") -> dict:
    groups = [values.dropna().to_numpy() for _, values in df.groupby(group_col, observed=True)[value_col]]
    stat, p_value = stats.kruskal(*groups)
    return {"variavel": group_col, "estatistica": stat, "p_valor": p_value}


def eta_squared_kruskal(h_stat: float, k: int, n: int) -> float:
    """Eta-quadrado para Kruskal-Wallis: mede proporção da variância
    explicada pela variável de grupo (0 = nenhum efeito, 1 = efeito total)."""
    return (h_stat - k + 1) / (n - k)


def dunn_posthoc(df: pd.DataFrame, group_col: str, value_col: str = "nps_score") -> pd.DataFrame:
    temp = df[[group_col, value_col]].dropna().copy()
    temp["rank"] = stats.rankdata(temp[value_col])
    groups = list(temp[group_col].dropna().unique())
    n_total = len(temp)

    tie_counts = temp[value_col].value_counts()
    tie_correction = 1 - ((tie_counts**3 - tie_counts).sum() / (n_total**3 - n_total))
    rank_var = (n_total * (n_total + 1) / 12) * tie_correction

    rows = []
    for i, group_a in enumerate(groups):
        for group_b in groups[i + 1:]:
            ranks_a = temp.loc[temp[group_col] == group_a, "rank"]
            ranks_b = temp.loc[temp[group_col] == group_b, "rank"]
            z_score = (ranks_a.mean() - ranks_b.mean()) / np.sqrt(rank_var * (1 / len(ranks_a) + 1 / len(ranks_b)))
            p_value = 2 * stats.norm.sf(abs(z_score))
            rows.append({
                "grupo_a": group_a,
                "grupo_b": group_b,
                "z": z_score,
                "p_valor": p_value,
            })

    result = pd.DataFrame(rows)
    if not result.empty:
        result["p_ajustado_bh"] = multipletests(result["p_valor"], method="fdr_bh")[1]
        result = result.sort_values("p_ajustado_bh")
    return result


kruskal_region = kruskal_test(working_df, "customer_region")
kruskal_age = kruskal_test(working_df, "age_group")

n_total = len(working_df)
n_regioes = working_df["customer_region"].nunique()
n_faixas = working_df["age_group"].nunique()

eta2_regiao = eta_squared_kruskal(kruskal_region["estatistica"], n_regioes, n_total)
eta2_faixa = eta_squared_kruskal(kruskal_age["estatistica"], n_faixas, n_total)

print("Kruskal-Wallis — Região:", kruskal_region)
print("Kruskal-Wallis — Faixa etária:", kruskal_age)
print(f"Eta-quadrado (região): {eta2_regiao:.4f}")
print(f"Eta-quadrado (faixa etária): {eta2_faixa:.4f}")
print("Referência: η² < 0.01 = efeito negligenciável | 0.01–0.06 = pequeno | > 0.14 = grande")

if kruskal_region["p_valor"] < 0.05:
    print("\nPost-hoc Dunn — Região")
    display(dunn_posthoc(working_df, "customer_region").round(5))
else:
    print("\nRegião: post-hoc Dunn não aplicado, pois o teste global não foi significativo.")

if kruskal_age["p_valor"] < 0.05:
    print("\nPost-hoc Dunn — Faixa etária")
    display(dunn_posthoc(working_df, "age_group").round(5))
else:
    print("\nFaixa etária: post-hoc Dunn não aplicado, pois o teste global não foi significativo.")
```

A análise pós-hoc indica que a maioria das comparações entre faixas etárias não apresenta diferença estatisticamente significativa após ajuste para múltiplos testes.

A única diferença significativa observada ocorre entre os grupos de 35–44 e 25–34 anos (p_ajustado = 0,03), indicando NPS inferior no grupo de 35–44. No entanto, essa diferença é isolada e de baixa magnitude prática.

De forma geral, os resultados sugerem que a faixa etária não é um driver relevante de satisfação nesta base.

### 5.3 NPS e recompra em 30 dias

- **H₀:** clientes que recompraram e clientes que não recompraram apresentam distribuição semelhante de NPS.
- **H₁:** clientes que recompraram apresentam distribuição de NPS diferente dos que não recompraram.

O Mann-Whitney U compara as distribuições sem assumir normalidade. O rank-biserial `r` complementa o p-valor com tamanho de efeito, mostrando se a diferença é apenas estatisticamente detectável ou também relevante para o negócio.

Quando a diferença é significativa e o tamanho de efeito é material, a leitura gerencial é direta: recompra em 30 dias funciona como um marcador comportamental associado à satisfação percebida. Ainda assim, ela deve ser usada com cuidado na modelagem, pois a disponibilidade temporal da variável precisa ser compatível com o momento em que o score preditivo será gerado.


```python
def rank_biserial_from_mannwhitney(u_stat: float, n1: int, n2: int) -> float:
    return (2 * u_stat) / (n1 * n2) - 1


repeat_yes = working_df.loc[working_df["repeat_purchase_30d"] == 1, "nps_score"]
repeat_no = working_df.loc[working_df["repeat_purchase_30d"] == 0, "nps_score"]

u_stat, p_value = stats.mannwhitneyu(repeat_yes, repeat_no, alternative="two-sided")
rank_biserial_r = rank_biserial_from_mannwhitney(u_stat, len(repeat_yes), len(repeat_no))

repeat_purchase_test = pd.DataFrame([{
    "grupo_1": "Recomprou em 30 dias",
    "n_grupo_1": len(repeat_yes),
    "media_grupo_1": repeat_yes.mean(),
    "mediana_grupo_1": repeat_yes.median(),
    "grupo_2": "Não recomprou em 30 dias",
    "n_grupo_2": len(repeat_no),
    "media_grupo_2": repeat_no.mean(),
    "mediana_grupo_2": repeat_no.median(),
    "u_stat": u_stat,
    "p_valor": p_value,
    "rank_biserial_r": rank_biserial_r,
}])

repeat_purchase_test.round(5)
```

Os resultados do teste de Mann-Whitney indicam uma diferença estatisticamente significativa entre os grupos (p-valor ≈ 0), levando à rejeição da hipótese nula.

Clientes que recompraram em até 30 dias apresentam NPS médio de 9,01 (mediana 9,0), enquanto clientes que não recompraram apresentam NPS médio de 3,94 (mediana 4,1). Essa diferença é expressiva e consistente entre média e mediana.

O tamanho de efeito, medido pelo rank-biserial (r = 1,0), indica separação praticamente completa entre os grupos, evidenciando que a recompra está fortemente associada à satisfação percebida.

Do ponto de vista de negócio, a recompra em 30 dias se configura como um forte marcador comportamental de satisfação: clientes promotores tendem a retornar, enquanto detratores dificilmente recompõem relacionamento no curto prazo.

Apesar disso, seu uso na modelagem deve ser feito com cautela, pois a variável pode não estar disponível no momento do scoring preditivo, dependendo do ponto da jornada em que o modelo será aplicado.

### 5.4 Associação entre categoria NPS e região

- **H₀:** a distribuição de Detratores, Neutros e Promotores é independente da região.
- **H₁:** existe associação entre categoria NPS e região.

O teste qui-quadrado avalia associação entre duas variáveis categóricas. O Cramér's V complementa a análise indicando força de associação. Se o p-valor não for significativo, a decisão mais prudente é priorizar fatores operacionais diretamente acionáveis antes de segmentações regionais isoladas. Se for significativo, a região pode entrar como critério de investigação adicional, mas ainda deve ser interpretada junto com atrasos, atendimento e reclamações.


```python
chi2_table = pd.crosstab(working_df["nps_category"], working_df["customer_region"]).reindex(category_order)
chi2_stat, chi2_p_value, chi2_dof, chi2_expected = stats.chi2_contingency(chi2_table)

n_total = chi2_table.to_numpy().sum()
min_dim = min(chi2_table.shape) - 1
cramers_v = np.sqrt(chi2_stat / (n_total * min_dim)) if min_dim > 0 else np.nan

print(f"Qui-quadrado: {chi2_stat:.4f}")
print(f"p-valor: {chi2_p_value:.4f}")
print(f"graus de liberdade: {chi2_dof}")
print(f"Cramér's V: {cramers_v:.4f}")
display(chi2_table)
```

A análise da distribuição de categorias de NPS por região indica proporções bastante semelhantes entre os grupos. Em todas as regiões, observa-se predominância de detratores, com participação consistente de neutros e baixa proporção de promotores.

Esse padrão homogêneo sugere ausência de variação relevante na experiência do cliente entre regiões. Assim, mesmo que o teste qui-quadrado identifique significância estatística (potencialmente influenciada pelo tamanho da amostra), a força da associação tende a ser baixa.

Do ponto de vista de negócio, isso indica que a região não é um fator determinante da satisfação nesta base. A insatisfação é generalizada e não concentrada em mercados específicos.

Dessa forma, a priorização de ações deve focar em fatores operacionais — como atraso, atendimento e reclamações — que demonstraram maior impacto direto no NPS, em vez de segmentações geográficas isoladas.

# **6. Preparação dos dados para modelagem**

A etapa de modelagem deve refletir o cenário real de produção: prever o NPS utilizando apenas variáveis disponíveis antes da coleta da pesquisa.

Dessa forma, são removidas as seguintes variáveis:

- `csat_internal_score`, por ser coletado simultaneamente ao NPS e caracterizar vazamento de informação (data leakage);
- `customer_id` e `order_id`, por serem identificadores únicos sem poder preditivo generalizável;
- `nps_score` e `nps_category`, conforme o tipo de tarefa, para evitar o uso da própria variável-alvo como entrada do modelo.

Essa separação é essencial para garantir a validade da modelagem. Modelos com data leakage tendem a apresentar métricas artificialmente elevadas, mas não se sustentam em produção, onde essas informações não estão disponíveis no momento do scoring.

```python
LEAKAGE_COLUMNS = ["csat_internal_score", "repeat_purchase_30d"]
IDENTIFIER_COLUMNS = ["customer_id", "order_id"]
TARGET_COLUMNS = ["nps_score", "nps_category"]

feature_columns = [
    column
    for column in working_df.columns
    if column not in LEAKAGE_COLUMNS + IDENTIFIER_COLUMNS + TARGET_COLUMNS
]

X = working_df[feature_columns].copy()
y_regression = working_df["nps_score"].copy()
y_classification = working_df["nps_category"].copy()

numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

print("Features numéricas:", numeric_features)
print("Features categóricas:", categorical_features)
print("Colunas removidas por governança preditiva:", LEAKAGE_COLUMNS + IDENTIFIER_COLUMNS + TARGET_COLUMNS)
```

```python
numeric_preprocessor = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_preprocessor = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_preprocessor, numeric_features),
    ("cat", categorical_preprocessor, categorical_features),
])
```

### 6.1 Multicolinearidade entre variáveis de entrega

`delivery_time_days` e `delivery_delay_days` podem carregar informação relacionada, mas não necessariamente redundante. O tempo total de entrega mede a duração da jornada logística, enquanto o atraso mede o descumprimento da promessa ao cliente.

A correlação e o VIF (Variance Inflation Factor — indicador que mede o quanto uma variável pode ser explicada por outras do modelo) ajudam a avaliar se há redundância entre as features. Valores elevados de VIF indicam multicolinearidade, podendo gerar instabilidade nos coeficientes do modelo.

```python
delivery_cols = ["delivery_time_days", "delivery_delay_days"]
delivery_corr = working_df[delivery_cols].corr()

vif_input = working_df[delivery_cols].dropna().copy()
vif_table = pd.DataFrame({
    "feature": delivery_cols,
    "VIF": [variance_inflation_factor(vif_input.values, i) for i in range(len(delivery_cols))],
})

display(delivery_corr.round(3))
display(vif_table.round(3))
```

Como regra prática, `VIF > 5` indica atenção e `VIF > 10` sugere multicolinearidade severa.

Na base analisada, a correlação entre `delivery_time_days` e `delivery_delay_days` é praticamente nula (-0,007), indicando ausência de relação linear entre as variáveis. Os valores de VIF (~2,3 para ambas) permanecem bem abaixo dos limiares críticos, confirmando que não há multicolinearidade relevante.

Dessa forma, as duas variáveis são mantidas no modelo, pois capturam aspectos distintos da experiência logística: o tempo total de entrega representa a duração da jornada, enquanto o atraso reflete o descumprimento da promessa ao cliente.

Caso os valores de VIF ultrapassassem os limiares estabelecidos, a priorização deveria recair sobre `delivery_delay_days`, por ser mais diretamente interpretável do ponto de vista de experiência do cliente.

# **7. Modelagem preditiva**

A modelagem é conduzida em duas abordagens complementares:

1. **Regressão:** previsão da nota individual de NPS.
2. **Classificação:** previsão da categoria de NPS, com atenção à separação entre detratores, neutros e promotores.

A avaliação é realizada por meio de validação cruzada com 5 folds, garantindo maior robustez e reduzindo a dependência de um único particionamento dos dados.

Adicionalmente, um conjunto de holdout é utilizado para validação final, com análise de matriz de confusão, métricas de classificação e importância das variáveis.

```python
regression_models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Random Forest Regressor": RandomForestRegressor(
        n_estimators=120,
        max_depth=8,
        min_samples_leaf=5,
        random_state=RANDOM_STATE,
        n_jobs=1,
    ),
    "Gradient Boosting Regressor": GradientBoostingRegressor(random_state=RANDOM_STATE),
}

classification_models = {
    "Logistic Regression balanced": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=RANDOM_STATE,
    ),
    "Random Forest Classifier balanced": RandomForestClassifier(
        n_estimators=120,
        max_depth=8,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=1,
    ),
    "Gradient Boosting Classifier": GradientBoostingClassifier(random_state=RANDOM_STATE),
}
```

```python
def evaluate_regression_cv(models: dict, X_data: pd.DataFrame, y_data: pd.Series) -> pd.DataFrame:
    cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
    rows = []

    for name, model in models.items():
        pipeline = Pipeline(steps=[("preprocess", preprocessor), ("model", model)])
        scores = cross_validate(
            pipeline,
            X_data,
            y_data,
            cv=cv,
            scoring={
                "r2": "r2",
                "rmse": "neg_root_mean_squared_error",
                "mae": "neg_mean_absolute_error",
            },
            n_jobs=1,
        )

        rows.append({
            "modelo": name,
            "R2_medio": scores["test_r2"].mean(),
            "R2_desvio": scores["test_r2"].std(),
            "RMSE_medio": -scores["test_rmse"].mean(),
            "RMSE_desvio": scores["test_rmse"].std(),
            "MAE_medio": -scores["test_mae"].mean(),
            "MAE_desvio": scores["test_mae"].std(),
        })

    return pd.DataFrame(rows).sort_values("RMSE_medio")


regression_cv_results = evaluate_regression_cv(regression_models, X, y_regression)
regression_cv_results.round(3)
```

As métricas de regressão devem ser interpretadas como desempenho obtido apenas com variáveis disponíveis antes da pesquisa, sem uso de `csat_internal_score`, garantindo ausência de vazamento de informação.

O modelo de Gradient Boosting apresentou o melhor desempenho, com R² médio de 0,553 e menor erro (RMSE de 1,67 e MAE de 1,32), seguido por Random Forest. Modelos lineares (Linear e Ridge) apresentaram desempenho ligeiramente inferior, porém próximo.

Os resultados indicam capacidade consistente de explicação do NPS a partir de variáveis operacionais, com cerca de 63% da variabilidade capturada pelo modelo.

Esse nível de desempenho é esperado em problemas de satisfação, nos quais parte relevante da variação depende de fatores subjetivos e não observáveis na base, como percepção individual, expectativa e contexto do cliente.

```python
def evaluate_classification_cv(models: dict, X_data: pd.DataFrame, y_data: pd.Series) -> pd.DataFrame:
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
    rows = []

    for name, model in models.items():
        pipeline = Pipeline(steps=[("preprocess", preprocessor), ("model", model)])
        scores = cross_validate(
            pipeline,
            X_data,
            y_data,
            cv=cv,
            scoring={
                "f1_macro": "f1_macro",
                "auc_ovr_weighted": "roc_auc_ovr_weighted",
            },
            n_jobs=1,
        )

        rows.append({
            "modelo": name,
            "F1_macro_medio": scores["test_f1_macro"].mean(),
            "F1_macro_desvio": scores["test_f1_macro"].std(),
            "AUC_ROC_OVR_medio": scores["test_auc_ovr_weighted"].mean(),
            "AUC_ROC_OVR_desvio": scores["test_auc_ovr_weighted"].std(),
        })

    return pd.DataFrame(rows).sort_values("F1_macro_medio", ascending=False)


classification_cv_results = evaluate_classification_cv(classification_models, X, y_classification)
classification_cv_results.round(3)
```

Na classificação, `class_weight='balanced'` é aplicado para mitigar o desbalanceamento entre classes, especialmente relevante em um cenário com predominância de detratores.

Os resultados indicam desempenho consistente entre os modelos, evidenciando bom equilíbrio na predição entre detratores, neutros e promotores. A classe **Neutro** permanece a mais desafiadora, por representar a fronteira entre satisfação e insatisfação.

O Random Forest apresentou o melhor resultado em F1 macro (0,51), enquanto os demais modelos apresentaram desempenho semelhante. As métricas de AUC (~0,87) indicam alta capacidade de separação entre as classes.

De forma geral, os modelos demonstram boa capacidade de classificação do NPS a partir de variáveis operacionais, sem dependência de informações com vazamento.

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X,
    y_regression,
    test_size=0.20,
    random_state=RANDOM_STATE,
)

best_regression_name = regression_cv_results.iloc[0]["modelo"]
best_regression_estimator = regression_models[best_regression_name]

best_regression_model = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("model", best_regression_estimator),
])

best_regression_model.fit(X_train_reg, y_train_reg)
y_pred_reg = best_regression_model.predict(X_test_reg)

regression_holdout_metrics = pd.DataFrame([{
    "modelo": best_regression_name,
    "R2": r2_score(y_test_reg, y_pred_reg),
    "RMSE": np.sqrt(mean_squared_error(y_test_reg, y_pred_reg)),
    "MAE": mean_absolute_error(y_test_reg, y_pred_reg),
}])

regression_holdout_metrics.round(3)
```

O modelo de Gradient Boosting manteve desempenho consistente no conjunto de teste, com R² de 0,545, próximo ao observado na validação cruzada . Essa estabilidade indica boa capacidade de generalização, sem evidência de overfitting relevante.

Os erros obtidos (RMSE de 1,69 e MAE de 1,36) são baixos considerando a escala do NPS (0 a 10), indicando que o modelo consegue estimar a nota com boa precisão média.

Do ponto de vista de negócio, o modelo é capaz de capturar padrões relevantes da experiência do cliente a partir de variáveis operacionais, sendo adequado para identificação de risco de insatisfação, mesmo sem uso de variáveis com vazamento de informação.

```python
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X,
    y_classification,
    test_size=0.20,
    stratify=y_classification,
    random_state=RANDOM_STATE,
)

best_classifier_name = classification_cv_results.iloc[0]["modelo"]
best_classifier_estimator = classification_models[best_classifier_name]

best_classifier = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("model", best_classifier_estimator),
])

best_classifier.fit(X_train_clf, y_train_clf)
y_pred_clf = best_classifier.predict(X_test_clf)

print(f"Melhor classificador por F1 macro medio: {best_classifier_name}")
print(classification_report(y_test_clf, y_pred_clf, digits=3))

if hasattr(best_classifier.named_steps["model"], "predict_proba"):
    y_proba_clf = best_classifier.predict_proba(X_test_clf)
    auc_roc_ovr = roc_auc_score(
        y_test_clf,
        y_proba_clf,
        multi_class="ovr",
        average="weighted",
        labels=best_classifier.named_steps["model"].classes_,
    )
    print("AUC-ROC OVR ponderado:", round(auc_roc_ovr, 3))

cm = confusion_matrix(y_test_clf, y_pred_clf, labels=category_order)
cm_pct = cm / cm.sum(axis=1, keepdims=True) * 100

fig, ax = plt.subplots(figsize=(7.4, 6.2))
annotations = np.array([[f"{cm[i, j]}\n{cm_pct[i, j]:.1f}%" for j in range(cm.shape[1])] for i in range(cm.shape[0])])
sns.heatmap(
    cm_pct,
    annot=annotations,
    fmt="",
    cmap="Blues",
    cbar_kws={"label": "% dentro da classe real"},
    linewidths=0.8,
    linecolor="white",
    xticklabels=category_order,
    yticklabels=category_order,
    ax=ax,
)
ax.set_title(f"Matriz de confusao - {best_classifier_name}", loc="left", fontsize=15, weight="bold")
ax.text(0, 1.06, "Cada celula mostra quantidade e percentual dentro da classe real.", transform=ax.transAxes, fontsize=11, color="#52616F")
ax.set_xlabel("Predito")
ax.set_ylabel("Real")
plt.tight_layout()
show_save_close(fig, "11_matriz_confusao_classificador.png")
```

O modelo **Random Forest Classifier balanced** foi o melhor classificador pelo critério de **F1 macro médio** .

No conjunto de teste, o modelo apresentou **acurácia de 0,798**, **F1 macro de 0,419** e **F1 ponderado de 0,805**. Esses resultados indicam um desempenho geral razoável, mas também evidenciam que a classificação entre as três categorias de NPS ainda é desafiadora, principalmente por causa do forte desbalanceamento da base.

A performance é mais robusta para a classe **Detrator**, com **F1-score de 0,908** e **recall de 0,889**, demonstrando boa capacidade de identificar clientes com maior risco de insatisfação — principal objetivo do modelo do ponto de vista de negócio.

As classes **Neutro** e **Promotor** apresentam desempenho inferior, com F1-score de **0,348** e **0**, respectivamente. Isso indica que o modelo ainda tem dificuldade para separar clientes em zonas intermediárias ou positivas de satisfação, especialmente pela menor quantidade de exemplos dessas classes no conjunto de teste.

A remoção de `repeat_purchase_30d` reduziu parte do poder preditivo, mas tornou o modelo mais coerente com o objetivo do desafio, evitando o uso de uma variável potencialmente indisponível no momento real da previsão.

De forma geral, o modelo deve ser interpretado como uma ferramenta de **alerta de risco de detratação**, e não como um classificador perfeito das três categorias de NPS. Sua principal utilidade prática está em priorizar clientes com sinais operacionais críticos para ações preventivas de CX, logística e atendimento.

### 7.1 Calibração de probabilidades e curvas Precision-Recall

A calibração de probabilidades ajusta as saídas do modelo para que representem melhor a frequência real observada. Em um contexto de CX, isso é essencial para que scores de risco sejam confiáveis na definição de prioridades de atendimento.

Já as curvas Precision-Recall são especialmente relevantes em cenários desbalanceados, pois evidenciam o trade-off entre capturar mais detratores (recall) e evitar falsos positivos (precisão).

Essa análise permite definir thresholds operacionais de forma mais informada, balanceando custo de intervenção e risco de não tratar clientes insatisfeitos.

```python
from sklearn.calibration import CalibratedClassifierCV

calibrated_clf = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("model", CalibratedClassifierCV(
        best_classifier_estimator,
        method="isotonic",
        cv=5,
    )),
])
calibrated_clf.fit(X_train_clf, y_train_clf)
y_pred_calibrated = calibrated_clf.predict(X_test_clf)

print("=== Classificador calibrado ===")
print(classification_report(y_test_clf, y_pred_calibrated, digits=3))
print(f"F1 macro (calibrado): {f1_score(y_test_clf, y_pred_calibrated, average='macro'):.3f}")
```

### 7.2 Ajuste de threshold para a classe Neutro

A classe **Neutro** representa uma zona intermediária entre satisfação e insatisfação, sendo naturalmente mais difícil de classificar. Em vez de depender apenas da decisão padrão do modelo, é possível ajustar o threshold da probabilidade dessa classe para controlar o trade-off entre precisão e recall.

Reduzir o threshold aumenta o recall de Neutros, capturando mais clientes nessa zona de atenção, ao custo de maior confusão com outras classes. Já aumentar o threshold melhora a precisão, mas reduz a cobertura.

Do ponto de vista de negócio, esse ajuste permite tratar clientes Neutros de forma estratégica: seja ampliando a identificação de clientes potencialmente insatisfeitos para ações preventivas, ou restringindo a classificação para focar apenas nos casos mais confiáveis. A escolha do threshold deve, portanto, refletir o custo operacional e o objetivo da estratégia de experiência do cliente.

```python
def predict_with_neutral_threshold(model: Pipeline, X_input: pd.DataFrame, neutral_threshold: float = 0.25) -> np.ndarray:
    if not hasattr(model.named_steps["model"], "predict_proba"):
        raise ValueError("O modelo selecionado não fornece predict_proba.")

    classes = model.named_steps["model"].classes_
    proba = model.predict_proba(X_input)
    neutral_idx = np.where(classes == "Neutro")[0][0]

    default_pred = classes[np.argmax(proba, axis=1)]
    adjusted_pred = default_pred.copy()
    adjusted_pred[proba[:, neutral_idx] >= neutral_threshold] = "Neutro"
    return adjusted_pred


if hasattr(best_classifier.named_steps["model"], "predict_proba") and "Neutro" in best_classifier.named_steps["model"].classes_:
    threshold_results = []
    for threshold in [0.20, 0.25, 0.30, 0.35, 0.40, 0.45]:
        adjusted_pred = predict_with_neutral_threshold(best_classifier, X_test_clf, neutral_threshold=threshold)
        precision, recall, f1, support = precision_recall_fscore_support(
            y_test_clf,
            adjusted_pred,
            labels=category_order,
            zero_division=0,
        )
        threshold_results.append({
            "threshold_neutro": threshold,
            "F1_macro": f1_score(y_test_clf, adjusted_pred, average="macro"),
            "precision_neutro": precision[1],
            "recall_neutro": recall[1],
            "f1_neutro": f1[1],
        })

    threshold_results_df = pd.DataFrame(threshold_results)
    display(threshold_results_df.round(3))
else:
    print("Ajuste de threshold não aplicado porque o classificador selecionado não expõe probabilidades compatíveis.")
```

A análise de sensibilidade do threshold para a classe **Neutro** evidencia um trade-off claro entre **recall** e **precisão**.

Thresholds mais baixos, entre **0,20 e 0,30**, aumentam fortemente o recall da classe Neutro, chegando a **0,964** no threshold 0,20. Isso significa que o modelo captura quase todos os clientes neutros, porém com baixa precisão, indicando maior confusão com outras classes.

À medida que o threshold aumenta, a precisão melhora, saindo de **0,2** em 0,20 para **0,281** em 0,40. Em contrapartida, o recall cai de forma relevante, chegando a **0,446** em 0,40 e **0,429** em 0,45.

Com base nos resultados, o melhor equilíbrio ocorre em torno de **0,45**, onde o modelo apresenta o maior **F1 macro (0,419)** e a melhor precisão para Neutro (**0,293**), mantendo ainda um recall moderado de **0,429**. Já o threshold **0,30** apresenta o melhor F1 específico para Neutro (**0,35**), com maior capacidade de capturar clientes neutros, mas menor desempenho global.

Do ponto de vista de negócio, a escolha depende da estratégia de CX. Se o objetivo for capturar mais clientes em zona intermediária de satisfação, mesmo com mais falsos positivos, o threshold **0,30** é mais adequado. Se o objetivo for maior equilíbrio geral entre as classes, o threshold **0,40** é a escolha mais consistente.

Assim, o threshold recomendado para o modelo final é **0,40**, por apresentar o melhor desempenho global, sem reduzir excessivamente a capacidade de identificação da classe Neutro.

### 7.4 Interpretabilidade com SHAP Values

A importância baseada em impureza (Gini), padrão de modelos como Random Forest e Gradient Boosting, tem viés estrutural para variáveis de alta cardinalidade: mesmo variáveis fracamente preditivas podem parecer relevantes por oferecerem mais pontos de divisão.

O **SHAP** (SHapley Additive exPlanations) corrige esse viés ao calcular a contribuição marginal de cada variável para cada predição individualmente, com base na teoria de jogos cooperativos. O resultado é uma medida de importância livre de viés de cardinalidade e interpretável em nível de observação.

Abaixo, o modelo de regressão (Gradient Boosting) é analisado via SHAP, permitindo comparar com a importância por impureza calculada anteriormente para o Random Forest Classifier.

```python
try:
    import shap
except ImportError:
    import subprocess
    subprocess.run([__import__("sys").executable, "-m", "pip", "install", "shap>=0.42"], check=True)
    import shap

# Preprocessar dados de teste (TreeExplainer opera sobre dados já transformados)
X_test_prep = best_regression_model.named_steps["preprocess"].transform(X_test_reg)
raw_feat_names = best_regression_model.named_steps["preprocess"].get_feature_names_out()
shap_labels = [
    LABEL_MAP.get(n.replace("num__", "").replace("cat__", ""), n.replace("num__", "").replace("cat__", ""))
    for n in raw_feat_names
]
X_shap = pd.DataFrame(X_test_prep, columns=shap_labels)

gb_explainer = shap.TreeExplainer(best_regression_model.named_steps["model"])
shap_vals = gb_explainer.shap_values(X_test_prep)

# Plot 1 — importância global (barras)
shap.summary_plot(shap_vals, X_shap, plot_type="bar", show=False)
fig_shap_bar = plt.gcf()
fig_shap_bar.set_size_inches(10.5, 7.5)
plt.gca().set_title(
    "Importância Global das Features — SHAP (Gradient Boosting Regressor)",
    loc="left", fontsize=13, weight="bold",
)
plt.tight_layout()
show_save_close(fig_shap_bar, "14_shap_summary_bar.png")

# Plot 2 — dependência de atraso na entrega
delay_label = "Atraso na entrega (dias)"
delay_idx = shap_labels.index(delay_label) if delay_label in shap_labels else 0
shap.dependence_plot(delay_idx, shap_vals, X_shap, show=False)
fig_shap_dep = plt.gcf()
fig_shap_dep.set_size_inches(10.5, 5.5)
ax_dep = fig_shap_dep.axes[0]
ax_dep.set_title(
    "Efeito do Atraso na Entrega no NPS previsto (SHAP)",
    loc="left", fontsize=12, weight="bold",
)
clean_axis(ax_dep, grid_axis="y")
plt.tight_layout()
show_save_close(fig_shap_dep, "15_shap_dependence_delivery_delay.png")
```

A variável de maior impacto global é **atraso na entrega** (`delivery_delay_days`), seguida por **número de reclamações** (`complaints_count`), **tempo de resolução** (`resolution_time_days`) e **contatos com atendimento** (`customer_service_contacts`). Esse resultado reforça que o NPS previsto é fortemente influenciado por fricções logísticas e de atendimento, e não por características demográficas ou regionais.

O gráfico de dependência de `delivery_delay_days` mostra um comportamento bastante claro: entregas sem atraso, ou com atraso negativo/antecipação, contribuem positivamente para o NPS previsto. À medida que o atraso aumenta, o valor SHAP se torna progressivamente negativo, indicando redução relevante na previsão de satisfação.

Esse padrão confirma o ponto de ruptura identificado na EDA: pequenos atrasos já reduzem a percepção do cliente, enquanto atrasos maiores produzem impacto fortemente negativo no NPS previsto.

Do ponto de vista de negócio, os resultados indicam que as principais alavancas de melhoria estão em reduzir atrasos, prevenir reclamações, acelerar a resolução de problemas e melhorar a eficiência do atendimento. Como essas variáveis são operacionais e acionáveis, o modelo se torna mais adequado para apoiar ações preventivas de CX, logística e atendimento.

### 7.6 Análise da Classe Neutro

A classe **Neutro** (NPS 7–8) apresenta F1-score de 0,57 — o menor do modelo de classificação. Neutros representam clientes em zona de transição: uma experiência ligeiramente pior os empurra para Detratores; uma melhoria os converte em Promotores.

Esta subseção investiga para onde o modelo erra: os Neutros são confundidos mais frequentemente com Detratores ou com Promotores? E quais características operacionais diferenciam os grupos mal classificados?

```python
y_pred_clf_s = pd.Series(y_pred_clf, index=y_test_clf.index, name="nps_predicted")

neutro_df = (
    pd.DataFrame({"nps_actual": y_test_clf, "nps_predicted": y_pred_clf_s})
    .join(
        working_df[["delivery_delay_days", "complaints_count", "customer_service_contacts"]],
        how="left",
    )
)

neutro_actual = neutro_df[neutro_df["nps_actual"] == "Neutro"].copy()
neutro_actual["grupo"] = neutro_actual["nps_predicted"].map({
    "Neutro": "Neutro correto",
    "Detrator": "Neutro → Detrator",
    "Promotor": "Neutro → Promotor",
})

n_neutros = len(neutro_actual)
contagem = neutro_actual["grupo"].value_counts()
print(f"Total de Neutros reais no conjunto de teste: {n_neutros}")
for grp, cnt in contagem.items():
    print(f"  {grp}: {cnt} ({cnt / n_neutros:.1%})")
print()

feats_comp = ["delivery_delay_days", "complaints_count", "customer_service_contacts"]
labels_comp = {
    "delivery_delay_days": "Atraso (dias)",
    "complaints_count": "Reclamações",
    "customer_service_contacts": "Contatos SAC",
}

display(
    neutro_actual.groupby("grupo")[feats_comp]
    .agg(["mean", "median"])
    .round(2)
)

grupos_ordem = ["Neutro correto", "Neutro → Detrator", "Neutro → Promotor"]
grupo_colors = [
    EXECUTIVE_COLORS["Neutro"],
    EXECUTIVE_COLORS["Detrator"],
    EXECUTIVE_COLORS["Promotor"],
]

fig, axes = plt.subplots(1, 3, figsize=(15.0, 5.5))
for i, feat in enumerate(feats_comp):
    ax = axes[i]
    data_groups = [
        neutro_actual.loc[neutro_actual["grupo"] == g, feat].dropna().values
        for g in grupos_ordem
    ]
    bp = ax.boxplot(
        data_groups,
        labels=["Correto", "→ Detrator", "→ Promotor"],
        patch_artist=True,
        medianprops=dict(color="#1F2D3D", linewidth=2),
        whiskerprops=dict(color="#6C757D"),
        capprops=dict(color="#6C757D"),
        flierprops=dict(marker="o", markersize=3, alpha=0.4),
    )
    for patch, color in zip(bp["boxes"], grupo_colors):
        patch.set_facecolor(color)
        patch.set_alpha(0.55)
    ax.set_title(labels_comp[feat], fontsize=11, weight="bold")
    ax.set_ylabel(labels_comp[feat])
    clean_axis(ax, grid_axis="y")

fig.suptitle("Perfil Operacional dos Neutros Mal Classificados", fontsize=14, weight="bold", y=1.03)
plt.tight_layout()
show_save_close(fig, "18_neutros_mal_classificados.png")
```

Neutros classificados como **Promotores** têm perfil operacional mais limpo, indicando que a nota 7–8 pode ser um teto subjetivo para clientes satisfeitos mas não entusiastas.

# **8. Aplicação em produção**

A aplicação prática do modelo deve ser estruturada como um pipeline de scoring operacional, integrado ao fluxo de jornada do cliente:

1. O pedido é concluído e os eventos da jornada logística e de atendimento são consolidados.
2. Variáveis operacionais disponíveis antes da pesquisa são capturadas, como atraso, tentativas de entrega, contatos com atendimento, reclamações, tempo de resolução, valor do pedido, frete, região e tempo de relacionamento.
3. O pipeline de preprocessamento aplica imputação, padronização e codificação das variáveis.
4. O modelo gera a previsão de categoria de NPS e um score de risco de detratação.
5. Clientes com maior risco são direcionados para uma fila de atuação proativa de CX.
6. A pesquisa NPS posterior é utilizada para validação e mensuração do impacto das intervenções.

Esse desenho garante aderência ao cenário real de produção e evita o principal risco metodológico do problema: o uso de variáveis com vazamento de informação. O modelo deve operar antes da coleta do NPS, não após.

### Estratégia de uso do modelo

Para fins operacionais, a abordagem de **classificação com threshold ajustado para a classe Detrator** é a mais adequada, pois responde diretamente à pergunta de negócio: *quais clientes devem ser priorizados para intervenção?*

A regressão pode ser utilizada como camada complementar para ranqueamento dentro da fila, permitindo priorizar clientes com maior risco relativo.

Dessa forma, a arquitetura recomendada é:

- Classificação → triagem (intervir ou não)
- Regressão → priorização (quem tratar primeiro)

### Critérios de monitoramento e retreinamento

Para garantir a sustentabilidade do modelo em produção, recomenda-se monitorar continuamente três dimensões:

| Dimensão | Métrica de alerta | Ação recomendada |
|---|---|---|
| Drift de dados | PSI > 0,2 nas features principais | Investigar mudanças operacionais e recalibrar |
| Drift de performance | Queda de F1 macro > 5 pp vs. baseline | Retreinar com dados mais recentes |
| Deriva de calibração | Hosmer-Lemeshow p < 0,05 | Recalibrar probabilidades |

A avaliação deve ser realizada mensalmente, com retreinamento semestral ou sob demanda, caso os limites de alerta sejam atingidos.

### Conexão com valor de negócio

A utilização do modelo permite transformar sinais operacionais em ações concretas de experiência do cliente, com foco em:

- Redução de detratores por meio de intervenções preventivas  
- Priorização eficiente de esforços de atendimento  
- Monitoramento contínuo da qualidade da operação  

Dessa forma, o modelo deixa de ser apenas um instrumento preditivo e passa a atuar como um mecanismo de gestão ativa da experiência do cliente.

# **10. Limitações e riscos**

Toda solução preditiva para NPS deve ser interpretada dentro de limites claros:

- **NPS é uma medida subjetiva:** parte relevante da variação depende de percepção individual, expectativa e contexto do cliente, não totalmente capturados nos dados.
- **Correlação não implica causalidade:** variáveis importantes no modelo indicam associação, mas não devem ser interpretadas automaticamente como causa direta.
- **Disponibilidade temporal das variáveis:** apenas informações conhecidas antes da pesquisa podem ser utilizadas, garantindo aplicabilidade real do modelo.
- **Ambiguidade da classe Neutro:** representa uma zona intermediária de difícil separação, exigindo ajustes de threshold e estratégia operacional específica.
- **Risco de drift:** mudanças operacionais, sazonais ou de comportamento do cliente podem alterar o desempenho do modelo ao longo do tempo.

A principal diretriz de governança é priorizar utilidade real sobre métricas aparentes. Modelos livres de vazamento de informação e aplicáveis em produção são mais valiosos do que modelos com desempenho artificialmente elevado, mas inviáveis no contexto operacional.

Em última instância, o modelo deve ser visto como um instrumento de apoio à decisão — e não como substituto da análise de negócio e da gestão da experiência do cliente.
