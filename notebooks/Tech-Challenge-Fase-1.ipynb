# **1. Entendimento do negócio (Business Understanding)**

Com o crescimento acelerado do e-commerce nacional, temos o cenário de uma empresa que passou a lidar com um volume cada vez maior de pedidos, entregas
e interações com clientes. Esse crescimento trouxe ganhos importantes de escala, mas também revelou desafios relevantes na experiência do cliente, especialmente refletidos na alta variabilidade do NPS entre diferentes perfis de consumidores. A área de Experiência do Cliente percebeu que, mesmo com
indicadores operacionais aparentemente semelhantes, alguns clientes se tornam promotores da marca, enquanto outros se tornam detratores. 

Essa diferença levanta uma questão central para o negócio: **quais fatores operacionais realmente influenciam a satisfação do cliente e como a empresa
pode agir de forma proativa para melhorar a experiência antes mesmo da aplicação da pesquisa de NPS?**

Atualmente, o NPS é coletado apenas após o encerramento da jornada de compra, o que limita a capacidade da empresa de antecipar problemas, priorizar ações corretivas e atuar de forma preventiva.

O desafio é a falta de previsibilidade. Atualmente, a empresa só descobre que falhou quando o cliente já teve uma experiência negativa.

## 1.1 O que é o NPS, como ele é medido e por que o NPS é importante?

Com base nesse problema, para seguirmos, primeiro precisamos entender: **O que é o NPS, como ele é medido e por que ele é importante?**

O Net Promoter Score é uma métrica de fidelidade e satisfação do cliente que mede a probabilidade de recomendação da sua empresa, produto ou serviço.

Ele é categorizado em três tipos:

🟢 Promotores (9-10): Clientes satisfeitos que indicam a marca.

🟡 Neutros (7-8): Clientes passivos, propensos à concorrência.

🔴 Detratores (0-6): Clientes insatisfeitos, propensos a falar mal da empresa.

O cálculo é feito subtraindo a porcentagem de detratores da porcentagem de promotores (excluindo os neutros):

NPS = % Promotores - % Detratores

Ele é um fator importante pois ajuda em tópicos como:

- Custo de Aquisição de Clientes (CAC): É mais barato manter um cliente do que conquistar um novo.

- Vantagem Competitiva: Em um mercado de produtos similares, a experiência de entrega, atendimento e pós-venda é um diferencial real.

- Validação de Processos: Pode revelar problemas que os números não mostram, como uma embalagem amassada ou uma experiência ruim que a logística padrão não percebe

## 1.2 Quais áreas poderiam se beneficiar desses insights

- **Logística:** Permite identificar transportadoras ou regiões específicas que causam mais insatisfação, mesmo entregando dentro do prazo. Isso ajuda a ajustar rotas, trocar parceiros logísticos ou melhorar o cuidado com o manuseio das encomendas.

- **Atendimento:** Priorizar o contato com clientes que enfrentaram problemas (como atrasos ou extravios) antes mesmo da pesquisa ser enviada.

- **Comercial:** Entender se a expectativa gerada pelo preço/oferta está alinhada com a percepção de valor na entrega.

- **Experiencia do cliente (UX/TI):** Ajustar o fluxo de comunicação para reduzir a ansiedade do cliente durante a espera (rastreio da entrega em tempo real).

## 1.3 Os Impactos do NPS

O NPS pode impactar em alguns pontos:

- **Taxa de Recompra e Ciclo de Vida (LTV):**
Clientes classificados como "Promotores" possuem um ciclo de vida maior e rentável, garantindo uma receita recorrente sem a necessidade de novos investimentos em mídia. No e-commerce, a margem de lucro real costuma se consolidar a partir da segunda ou terceira transação; portanto, um NPS baixo indica que a empresa está perdendo clientes prematuramente, tornando os investimentos em marketing ineficientes e incapazes de gerar crescimento sustentável.

- **Marketing Boca a Boca:**
O e-commerce é movido pela confiança. Clientes "Detratores" não apenas deixam de comprar, mas podem expor falhas em redes sociais e sites de reclamação, o que gera uma resistência maior em novos compradores e encarece o marketing. Por outro lado, os "Promotores" tornam-se um canal de marketing gratuito e de alta conversão: eles recomendam a marca espontaneamente, trazendo novos clientes com maior confiança.

- **Market Share:**
Embora a fatia de mercado possa ser expandida rapidamente através de promoções e escala logística, a liderança só é mantida através da fidelização. Empresas que sustentam um NPS elevado conseguem dominar o setor a longo prazo porque constroem comunidades de consumo sólidas. Essa lealdade cria uma barreira de proteção contra a concorrência, permitindo que a marca mantenha sua relevância mesmo diante de disputa de preços ou flutuações do mercado.

## 1.4 Indicadores que podem complementar a análise
Para melhorar o NPS, a empresa pode considerar aderir ou revisar alguns cenários:

- **CES (Customer Effort Score):** Mede o quão fácil foi para o cliente resolver um problema ou concluir a compra. Muitas vezes, um NPS baixo está ligado ao alto esforço.

- **OTD (On-Time Delivery):** Percentual de pedidos entregues exatamente no prazo. O atraso é o maior gerador de detratores no e-commerce.

- **First Contact Resolution (FCR):** Se o cliente precisou entrar em contato, o problema foi resolvido na primeira tentativa?

- **Churn Rate:** Cruzar a nota do NPS com o índice de abandono da base para validar a correlação entre insatisfação e perda real de receita.

- **SLA de Logística Inversa:** A facilidade de devolução é um fator crítico que pode transformar um potencial detrator em um promotor pela confiança gerada.

Assim, podemos usar os dados operacionais como um Sistema de Alerta: Se um pedido atrasou e o cliente já reclamou, a probabilidade de ele ser um detrator é alta. Agir antes de enviar a pesquisa é essencial para o crescimento da empresa.

### 2.3 Análise do NPS

```python
#Função Classificação de NPS
#Promotores (9-10) Neutros (7-8) Detratores (0-6)

def nps_class(dataset):
  if dataset['nps_score'] >= 9:
    return 'Promotor'
  elif dataset['nps_score'] >= 7:
    return 'Neutro'
  else:
    return 'Detrator' 
```

```python
#Adicionar a coluna de classificação de NPS ao dataset
dataset_nps = pd.DataFrame(dataset.apply(nps_class, axis=1), columns=['classe_nps'])
dataset_nps
```

```python
#Contar a quantidade de cada classificação de NPS
dataset_nps['classe_nps'].value_counts()
```

Aqui é possivel observar que aproximadamente 84% de todos os NPS são Detratores.

## 3.1 Criando categorias

```python
dataset_eda['repeat_purchase_30d_category'] = dataset_eda['repeat_purchase_30d'].map({
    0: 'Não',
    1: 'Sim'
})
dataset_eda.head()
```

Criamos variáveis derivadas para tornar a análise mais interpretável do ponto de vista de negócio.

As principais transformações incluem:

- **nps_category:** classifica o cliente em detrator, neutro ou promotor
- **is_delayed:** indica se houve atraso na entrega
- **had_complaint:** indica se houve registro de reclamação
- **had_support_contact:** indica se o cliente acionou o atendimento
- **is_repeat_customer:** identifica recompra em até 30 dias
- **multiple_delivery_attempts:** sinaliza mais de uma tentativa de entrega
- **age_group** e **tenure_group:** facilitam comparações por perfil de cliente

```python
def classify_nps(score: float) -> str:
    if score >= 9:
        return "Promotor"
    elif score >= 7:
        return "Neutro"
    return "Detrator"


dataset_eda["nps_category"] = dataset_eda["nps_score"].apply(classify_nps)
dataset_eda["is_delayed"] = (dataset_eda["delivery_delay_days"] > 0).astype(int)
dataset_eda["had_complaint"] = (dataset_eda["complaints_count"] > 0).astype(int)
dataset_eda["had_support_contact"] = (dataset_eda["customer_service_contacts"] > 0).astype(int)
dataset_eda["is_repeat_customer"] = (dataset_eda["repeat_purchase_30d"] > 0).astype(int)
dataset_eda["multiple_delivery_attempts"] = (dataset_eda["delivery_attempts"] > 1).astype(int)

dataset_eda["age_group"] = pd.cut(
    dataset_eda["customer_age"],
    bins=[18, 25, 35, 45, 55, 70],
    labels=["18-24", "25-34", "35-44", "45-54", "55+"],
    right=False
)

dataset_eda["tenure_group"] = pd.cut(
    dataset_eda["customer_tenure_months"],
    bins=[0, 12, 36, 60, 120],
    labels=["0-11m", "12-35m", "36-59m", "60m+"],
    right=False
)

dataset_eda.head()
```

## 3.3 Comparação de médias por categoria de NPS

Comparamos indicadores operacionais entre detratores, neutros e promotores para identificar diferenças na experiência do cliente.

```python
order = ["Detrator", "Neutro", "Promotor"]
metrics = [
    "delivery_delay_days",
    "customer_service_contacts",
    "complaints_count",
    "csat_internal_score"
]

titles = [
    "Atraso médio",
    "Contatos com atendimento",
    "Reclamações",
    "CSAT interno"
]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.ravel()

for ax, metric, title in zip(axes, metrics, titles):
    series = dataset_eda.groupby("nps_category")[metric].mean().reindex(order)

    sns.barplot(x=series.index, y=series.values, ax=ax)

    # adiciona valor no topo
    for i, v in enumerate(series.values):
        ax.text(i, v, f"{v:.2f}", ha="center", va="bottom", fontweight="bold")

    ax.set_title(title)
    ax.set_xlabel("")
    ax.set_ylabel("")

plt.tight_layout()
plt.show()
```

A comparação entre detratores, neutros e promotores evidencia um padrão consistente na experiência do cliente.

Clientes classificados como detratores apresentam:

- maior atraso médio (2.41 dias vs 0.72 em promotores)
- mais contatos com atendimento (1.63 vs 0.67)
- maior volume de reclamações (4.44 vs 2.27)
- maior tempo de resolução (5.69 dias vs 3.70)
- menor percepção de satisfação interna (CSAT: 2.54 vs 6.04)

Por outro lado, variáveis financeiras como valor do pedido e desconto apresentam pouca variação entre os grupos, indicando baixo impacto direto na satisfação.

## 3.4 Ponto de ruptura na experiência — atraso de entrega

Analisamos o impacto do atraso de entrega no NPS para identificar possíveis pontos de ruptura na experiência do cliente.

A pergunta central é:

## 4.1 Criando a pipeline para padronização e com os modelos:
- Regressão Linear
- Árvore de decisão

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

# Cria pipeline com duas etapas: padronização + regressão linear
pipeline_lin_reg = Pipeline([
    ('scaler', StandardScaler()),
    ('lin_reg', LinearRegression())
])

# Cria pipeline com duas etapas: padronização + regressão linear
pipeline_tree_reg = Pipeline([('tree_reg', DecisionTreeRegressor())])

pipeline_rf_reg = Pipeline([('rf_reg', RandomForestRegressor(
    n_estimators=100,      # número de árvores
    max_depth=None,        # profundidade máxima (None = cresce até não poder mais)
    random_state=42)
)])

pipeline_rf_class = Pipeline([('rf_class', RandomForestClassifier(
    n_estimators=200,
    random_state=42)
)])
```

## 4.2 Treino e validação do modelo de Regressão Linear

```python
# Treina o pipeline (fit aplica scaler e treina o modelo)
pipeline_lin_reg.fit(X_train, y_train)

# Faz previsões diretamente (o pipeline cuida da padronização do X_test)
y_pred = pipeline_lin_reg.predict(X_test)
```
